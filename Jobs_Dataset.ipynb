```python
import random
import pandas as pd
import numpy as np
```

```python
random.seed(42)
np.random.seed(42)

# --- Config ---
N_CANDIDATES = 1000
N_JOBS = 200

skills_pool = [
    "Python", "SQL", "Pandas", "NumPy", "Scikit-learn",
    "TensorFlow", "PyTorch", "Power BI", "Tableau",
    "Excel", "AWS", "Docker", "Git", "Spark"
]

locations = ["Tel Aviv", "Haifa", "Jerusalem", "Remote"]
education_levels = ["High School", "Bachelor", "Master", "PhD"]
```

```python
# --- Generate Candidates ---
def generate_candidate(candidate_id):
    skills = random.sample(skills_pool, random.randint(3, 6))
    return {
        "candidate_id": candidate_id,
        "skills": ",".join(skills),
        "experience_years": random.randint(0, 10),
        "education": random.choice(education_levels),
        "location": random.choice(locations),
        "salary_expectation": random.randint(12000, 30000)
    }

candidates = pd.DataFrame(
    [generate_candidate(i) for i in range(N_CANDIDATES)]
)
```

```python
# --- Generate Jobs ---
def generate_job(job_id):
    skills = random.sample(skills_pool, random.randint(3, 6))
    return {
        "job_id": job_id,
        "required_skills": ",".join(skills),
        "min_experience": random.randint(0, 8),
        "education_required": random.choice(education_levels),
        "location": random.choice(locations),
        "salary_offer": random.randint(14000, 32000)
    }

jobs = pd.DataFrame(
    [generate_job(i) for i in range(N_JOBS)]
)
```

```python
# --- Save CSV ---
candidates.to_csv("candidates.csv", index=False)
jobs.to_csv("jobs.csv", index=False)

print("Dataset generated successfully!")
print("Candidates:", len(candidates))
print("Jobs:", len(jobs))
```
